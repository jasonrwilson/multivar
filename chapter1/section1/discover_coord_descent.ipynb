# Discover Coordinate Descent

```python
#@title Find a Least Squares Line

beta_1 = 1.5 #@param {type:"number"}
beta_2 = 1.25 #@param {type:"number"}

import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [10,6]
plt.rcParams.update({'font.size': 14})

x = np.array([0,1,2,3,4])
y = np.array([2,3,3,5,7])

def sum_of_squares_error(beta_1, beta_2):
    return np.sum((beta_1 + beta_2*x - y)**2)


fig, axs = plt.subplots(1, 2)
fig.tight_layout(pad=3.0)

# Vary beta_1 while fixing beta_2
beta_1_arr = np.linspace(beta_1 - 0.05, beta_1 + 0.05, 51)

sse_beta_1 = [
    sum_of_squares_error(b1, beta_2)
    for b1 in beta_1_arr
]

# Vary beta_2 while fixing beta_1
beta_2_arr = np.linspace(beta_2 - 0.05, beta_2 + 0.05, 51)

sse_beta_2 = [
    sum_of_squares_error(beta_1, b2)
    for b2 in beta_2_arr
]


fig.suptitle(
    'SSE(' + str(beta_1) + ',' + str(beta_2) + ') = %g'
    % sum_of_squares_error(beta_1, beta_2)
)

axs[0].scatter(
    [beta_1],
    [sum_of_squares_error(beta_1, beta_2)],
    s=50,
    color='red'
)

axs[1].scatter(
    [beta_2],
    [sum_of_squares_error(beta_1, beta_2)],
    s=50,
    color='red'
)

axs[0].plot(beta_1_arr, sse_beta_1)
axs[1].plot(beta_2_arr, sse_beta_2)

axs[0].set_title('Fix the slope at ' + str(beta_2))
axs[1].set_title('Fix the intercept at ' + str(beta_1))

axs[0].set_xlabel('intercept')
axs[1].set_xlabel('slope')

axs[0].set_ylabel('sum of squared errors')
axs[1].set_ylabel('sum of squared errors')

plt.show()
```

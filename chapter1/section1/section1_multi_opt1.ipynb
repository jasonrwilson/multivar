# Section 1.1 : A Multivariable Optimization Problem : An Adhoc Approach

```python
#@title Find a Least Squares Line

beta_1 = 1.5 #@param {type:"number"}
beta_2 = 1.25 #@param {type:"number"}

import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = [6,4]
plt.rcParams.update({'font.size': 14})

x = np.array([0,1,2,3,4])
y = np.array([2,3,3,5,7])

def sum_of_squares_error(beta_1, beta_2):
    return np.sum((beta_1 + beta_2*x - y)**2)

plt.scatter(x, y,color='black',s=50)
plt.plot([0,4],
         [beta_1, beta_1 + 4*beta_2],
         color='maroon')

plt.title(
    f"f({beta_1},{beta_2}) = "
    f"{sum_of_squares_error(beta_1,beta_2):.3f}"
)
plt.show()
```
